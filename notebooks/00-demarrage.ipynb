## 1. Imports

```python
import pandas as pd
import json
from pathlib import Path

# Confort affichage
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 200)

DATA_DIR = Path('..') / 'data'
RAW = DATA_DIR / 'raw'
REF = DATA_DIR / 'reference'

print('Données dans :', RAW.resolve())
```

## 2. Chargement des fichiers CSV

```python
organizations = pd.read_csv(RAW / 'organizations.csv')
users = pd.read_csv(RAW / 'users.csv')
projects = pd.read_csv(RAW / 'projects.csv')
tasks = pd.read_csv(RAW / 'tasks.csv')
task_assignments = pd.read_csv(RAW / 'task_assignments.csv')
comments = pd.read_csv(RAW / 'comments.csv')
metrics = pd.read_csv(RAW / 'performance_metrics.csv')
promos = pd.read_csv(REF / 'promos.csv')

print('CSV chargés ✓')
```

## 3. Chargement du fichier JSON

```python
with open(RAW / 'activity_logs.json', 'r', encoding='utf-8') as f:
    activity_logs_raw = json.load(f)

activity_logs = pd.DataFrame(activity_logs_raw)
print(f'{len(activity_logs)} logs d\'activité chargés ✓')
```
